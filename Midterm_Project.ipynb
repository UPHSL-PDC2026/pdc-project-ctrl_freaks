```python
import kagglehub
import os
import pandas as pd

# Download dataset
path = kagglehub.dataset_download("likithagedipudi/linkedin-compatibility-dataset-50k-profiles")

# See files
print(os.listdir(path))

# Load dataset (replace filename!)
df = pd.read_csv(path + "/compatibility_pairs.csv")

# Count rows
print("Rows:", df.shape[0])
```

```python
def filter_data(data):
    results['filter'] = data[data['compatibility_score'] > 70]
```

```python
def average_score(data):
    results['avg'] = data['compatibility_score'].mean()
```

```python
def sort_data(data):
    results['sort'] = data.sort_values(by='compatibility_score', ascending=False)
```

```python
import time

start = time.time()

filtered = filter_data(df)
avg = average_score(df)
sorted_df = sort_data(df)

end = time.time()

print("Average Score:", avg)
print("Sequential Time:", end - start)
```

```python
def execute_task(func, data):
    return func(data)
```

```python
import multiprocessing
import time


if __name__ == "__main__":
    start = time.time()

    with multiprocessing.Pool(3) as pool:
        results = pool.starmap(
            execute_task,[(filter_data, df),(average_score, df),(sort_data, df)]
        )

    end = time.time()

    print("Parallel Time:", end - start)
```

```python
import threading
results = {}  # Dictionary to store parallel results

start_par = time.time()

t1 = threading.Thread(target=filter_data, args=(df,))
t2 = threading.Thread(target=average_score, args=(df,))
t3 = threading.Thread(target=sort_data, args=(df,))

t1.start()
t2.start()
t3.start()

t1.join()
t2.join()
t3.join()

end_par = time.time()
par_time = end_par - start_par

print("Parallel Execution Time (Threading):", par_time)
```

```python
print("Parallel Execution Time (Threading):", par_time)
print("Filtered Rows (>70):", len(results['filter']))
print("Average Compatibility Score:", results['avg'])
print("Top 5 Sorted Rows by Compatibility Score:")
print(results['sort'].head())
```
